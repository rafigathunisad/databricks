```sql
CREATE OR REPLACE TABLE electronics_cat.bronze.customers AS
SELECT
  customer_key,
  gender,
  name,
  city,
  state_code,
  state,
  zip_code,
  country,
  continent,
  birthday
FROM azure_blob_storage.customers;
```

```sql
CREATE OR REPLACE TABLE electronics_cat.bronze.products AS
SELECT
  product_key,
  product_name,
  brand,
  color,
  unit_cost_usd,
  unit_price_usd,
  subcategory_key,
  subcategory,
  category_key,
  category
FROM azure_blob_storage.products;

CREATE OR REPLACE TABLE electronics_cat.bronze.sales AS
SELECT
  order_number,
  line_item,
  order_date,
  delivery_date,
  customer_key,
  store_key,
  product_key,
  quantity,
  currency_code
FROM azure_blob_storage.sales;

CREATE OR REPLACE TABLE electronics_cat.bronze.stores AS
SELECT
  store_key,
  country,
  state,
  square_meters,
  open_date
FROM azure_blob_storage.stores;

CREATE OR REPLACE TABLE electronics_cat.bronze.exc_rate AS
SELECT
  date,
  currency,
  exchange
FROM azure_blob_storage.exc_rate;

CREATE OR REPLACE TABLE electronics_cat.bronze.data_dic AS
SELECT
  table,
  field,
  description
FROM azure_blob_storage.data_dic;



```
