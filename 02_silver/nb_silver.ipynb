```sql
CREATE OR REPLACE TABLE electronics_cat.silver.customers AS
SELECT
  CAST(customer_key AS INT) AS customerkey,
  LOWER(gender) AS gender,
  name,
  LOWER(city) AS city,
  state_code,
  state,
  zip_code,
  country,
  continent,

  
  COALESCE(
    try_to_date(birthday, 'M/d/yyyy'),
    try_to_date(birthday, 'd-M-yyyy'),
    try_to_date(birthday, 'dd-MM-yyyy')
  ) AS birthday

FROM electronics_cat.bronze.customers
WHERE customer_key IS NOT NULL;
```

```sql
CREATE OR REPLACE TABLE electronics_cat.silver.sales AS
SELECT
  order_number,
  line_item,

  COALESCE(
    try_to_date(order_date, 'M/d/yyyy'),
    try_to_date(order_date, 'd-M-yyyy'),
    try_to_date(order_date, 'dd-MM-yyyy')
  ) AS order_date,

  COALESCE(
    try_to_date(delivery_date, 'M/d/yyyy'),
    try_to_date(delivery_date, 'd-M-yyyy'),
    try_to_date(delivery_date, 'dd-MM-yyyy')
  ) AS delivery_date,

  CAST(customer_key AS INT) AS customerkey,
  COALESCE(CAST(store_key AS INT), -1) AS storekey,
  product_key,
  CAST(quantity AS INT) AS quantity,
  currency_code

FROM electronics_cat.bronze.sales;

```

```sql
CREATE OR REPLACE TABLE electronics_cat.silver.stores AS
SELECT
  store_key,
  country,
  state,
  square_meters,

  COALESCE(
    try_to_date(open_date, 'M/d/yyyy'),
    try_to_date(open_date, 'd-M-yyyy'),
    try_to_date(open_date, 'dd-MM-yyyy')
  ) AS open_date

FROM electronics_cat.bronze.stores;
```

```sql
CREATE OR REPLACE TABLE electronics_cat.silver.exc_rate AS
SELECT
  COALESCE(
    try_to_date(date, 'M/d/yyyy'),
    try_to_date(date, 'd-M-yyyy'),
    try_to_date(date, 'dd-MM-yyyy')
  ) AS date,
  currency,
  CAST(exchange AS DOUBLE) AS exchange
FROM electronics_cat.bronze.exc_rate;
```

```sql
CREATE OR REPLACE TABLE electronics_cat.silver.products AS
SELECT
  product_key AS productkey,
  product_name,
  brand,
  color,

  
  try_cast(
    REGEXP_REPLACE(unit_cost_usd, '[$, ]', '') 
    AS DOUBLE
  ) AS unit_cost_usd,

  
  try_cast(
    REGEXP_REPLACE(unit_price_usd, '[$, ]', '') 
    AS DOUBLE
  ) AS unit_price_usd,

  subcategory_key,
  subcategory,
  category_key,
  category

FROM electronics_cat.bronze.products;
```
